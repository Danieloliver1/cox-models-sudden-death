```python
# converter_wfdb_completo_para_parquet.py

import pandas as pd
import glob
import wfdb
import os
from tqdm import tqdm

print("--- INICIANDO SCRIPT DE CONVERSÃO DIRETA (WFDB para PARQUET) ---")
print("--- MODO: Sinal completo, sem limpeza ---")

# --- 1. CONFIGURAÇÃO ---
# ✅ MODIFIQUE AQUI: Caminho para a pasta com seus arquivos WFDB brutos (P*.hea, P*.dat, etc.)
path_raw_wfdb = r'D:\cox-models-sudden-death\01_Dataset\dados_ECG\Holter_ECG\P*'

# ✅ MODIFIQUE AQUI: Caminho para a pasta onde os arquivos .parquet serão salvos
path_output_parquet = r"D:\cox-models-sudden-death\01_Dataset\dados_ECG\dados_ECG_parquet"

# --- 2. LÓGICA DE CONVERSÃO ---

# Cria o diretório de saída se ele não existir
print(f"Diretório de saída: {path_output_parquet}")
os.makedirs(path_output_parquet, exist_ok=True)

# Encontra todos os arquivos base (ex: '.../P001')
arquivos_base = glob.glob(path_raw_wfdb)
lista_pacientes_base = sorted(list(set([os.path.splitext(f)[0] for f in arquivos_base])))

print(f"Encontrados {len(lista_pacientes_base)} pacientes para converter.")

sucessos = 0
falhas = 0

# Itera sobre cada paciente com uma barra de progresso
for base_path in tqdm(lista_pacientes_base, desc="Convertendo pacientes"):
    paciente_id = os.path.basename(base_path)
    try:
        # Lendo o registro WFDB completo (.hea e .dat)
        record = wfdb.rdrecord(base_path)
        p_signal = record.p_signal

        # Verifica se há pelo menos 3 canais
        if p_signal.shape[1] < 3:
            print(f"\nAVISO: Paciente {paciente_id} tem menos de 3 canais. Pulando...")
            falhas += 1
            continue

        # Seleciona os 3 primeiros canais do sinal bruto completo
        sinal_x = p_signal[:, 0]
        sinal_y = p_signal[:, 1]
        sinal_z = p_signal[:, 2]
        
        # Cria um DataFrame final com os dados brutos
        df_final = pd.DataFrame({
            "sinal_x": sinal_x,
            "sinal_y": sinal_y,
            "sinal_z": sinal_z
        })

        # Define o nome do arquivo de saída .parquet
        output_filename = os.path.join(path_output_parquet, f"{paciente_id}.parquet")
        
        # Salva o DataFrame no formato Parquet, que é rápido e eficiente
        df_final.to_parquet(output_filename, index=False)
        
        sucessos += 1

    except Exception as e:
        print(f"\nERRO ao processar o paciente {paciente_id}: {e}")
        falhas += 1
        continue

print("\n--- CONVERSÃO CONCLUÍDA ---")
print(f"✅ Sucessos: {sucessos}")
print(f"❌ Falhas: {falhas}")
```
